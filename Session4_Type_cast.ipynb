**Why We Need Type Casting**

```python
n1=100
n2=200
n1+n2
```

```python
n1=input('enter a n1:')  #'100'
n2=input('enter a n2:')  #'200'  because input() ALWAYS returns a string.
n1+n2  #'100' + '200' = '100200'  that is why we need type cast
```

**string concatination**

```python
n1=int(input('enter a n1:'))  # convert string'100' to integer 100
n2=int(input('enter a n2:'))
n1+n2

# This is called: String Concatenation (Joining strings together.)
```

```python
# Q: Why do we use int(input())?

# Answer:
# Because input() returns data in string format. 
# To perform numerical calculations, we convert the string into an integer using int().
```
